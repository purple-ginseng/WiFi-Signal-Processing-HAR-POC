```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# df = pd.read_csv("real_imag_csv/bfm_data_0807_jinwei_livetest1_20260708_155704.csv")
# df = pd.read_csv("real_imag_csv/bfm_data_0807_jinwei_livetest2_20260708_155926.csv")
# df = pd.read_csv("mean_phase_csv/bfm_data_0807_sidney_test2_20260708_150011.csv")
df = pd.read_csv("bfm_real_imag_csv/bfm_data_1007_sidney_test1_20260710_132842.csv")
```

```python
real_df = df[df.columns[df.columns.str.endswith('Real')]]
imag_df = df[df.columns[df.columns.str.endswith('Imag')]]
```

```python
mag_df = np.sqrt(np.pow(real_df.to_numpy(), 2) + np.pow(imag_df.to_numpy(), 2))
```

```python
pca = PCA(n_components = 1)
mag_pca = pca.fit_transform(mag_df)
```

```python
sns.lineplot(x = np.arange(len(mag_pca)), y = mag_pca.reshape(-1))
plt.show()
```
